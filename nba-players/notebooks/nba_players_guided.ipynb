## Setup and data preview

```python
from pathlib import Path
import runpy
import sys

import pandas as pd
from IPython.display import Image, display

for candidate in (Path.cwd(), *Path.cwd().parents):
    if (candidate / "common" / "data_loading.py").exists():
        PROJECT_ROOT = candidate
        break
else:
    raise FileNotFoundError("Could not locate the repository root.")

sys.path.insert(0, str(PROJECT_ROOT))
from common.data_loading import load_dataset

OUTPUT_DIR = PROJECT_ROOT / "nba-players" / "outputs"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

def run_exercise(script_name, output_name):
    """Run one canonical exercise and display its saved chart."""
    script_path = PROJECT_ROOT / "nba-players" / "exercises" / script_name
    print(f"Running: {script_path.relative_to(PROJECT_ROOT)}")
    runpy.run_path(str(script_path), run_name="__main__")
    chart_path = OUTPUT_DIR / output_name
    if not chart_path.is_file():
        raise FileNotFoundError(f"Expected chart was not created: {chart_path}")
    display(Image(filename=str(chart_path)))

_, raw_df = load_dataset(
    "nba-players/dataset/nba_dataset.csv",
    list(pd.read_csv(PROJECT_ROOT / "nba-players/dataset/nba_dataset.csv", nrows=0).columns),
)
analysis_df = raw_df.copy()
print(f"Rows: {len(analysis_df)}")
display(analysis_df.head())
```

The notebook uses an in-memory copy for exploration. The source CSV remains unchanged.
